checking the failure Lables



```python
print("\nFAILURE LABEL DISTRIBUTION ")
label_columns = [
    "Machine failure", "TWF", "HDF", "PWF", "OSF", "RNF"
]

for column in label_columns:
    if column in df.columns:
        print(f"\n{column}:")
        print(df[column].value_counts(dropna=False).sort_index())
```

Check expected categorical values

```python
if "Type" in df.columns:
    print(df["Type"].value_counts(dropna= False)) 
```

Check numeric columns for negative values

```python
numeric_column = df.select_dtypes(include="number")
for column in numeric_column.columns:
    negetive_count = (df[column] < 0).sum()
    if negetive_count > 0 :
        print(f"{column} : {negetive_count} negetive values ")
```

Check whether UDI is unique

```python
print("Rows : ", len(df))
print("Columns : ", len(df.columns))
print("Missing Cells : ", df.isnull().sum().sum())
print("duplicate rows :", df.duplicated().sum())

if "UDI" in df.columns:
    print("UDI is unique :", df["UDI"].is_unique)
```
